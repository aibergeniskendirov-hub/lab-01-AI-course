```python
for config in ["A", "B"]:
    total = 0

    for rank in ranks["rank_" + config]:
        if rank <= 10:
            total += 1 / rank

    mrr = total / len(ranks)
    print(config, "MRR@10 =", round(mrr, 6))
```

```python
import math

def wilson(hits, n):
    p = hits / n
    z = 1.96

    denominator = 1 + z**2 / n
    centre = (p + z**2 / (2 * n)) / denominator
    half = z * math.sqrt(
        p * (1 - p) / n + z**2 / (4 * n**2)
    ) / denominator

    return centre - half, centre + half

for config in ["A", "B"]:
    for k in [3, 5]:
        hits = (ranks["rank_" + config] <= k).sum()
        n = len(ranks)
        lower, upper = wilson(hits, n)

        print(
            f"{config}, k={k}: "
            f"recall={hits / n:.2%}, "
            f"lower={lower:.2%}, upper={upper:.2%}"
        )
```

```python
languages = [
    ("EN", 0.56, 26, 127, 275, 67),
    ("RU", 0.27, 43, 208, 451, 110),
    ("KK", 0.17, 50, 244, 528, 129)
]

for config in ["A", "B"]:
    for k in [3, 5]:
        average_cost = 0

        for lang, share, question, answer, chunk_a, chunk_b in languages:
            chunk = chunk_a if config == "A" else chunk_b
            input_tokens = 278 + question + k * chunk

            request_cost = (input_tokens * 1 + answer * 6) / 1000000
            average_cost += share * request_cost

        annual_cost = average_cost * 12000 * 365
        print(f"{config}, k={k}: {annual_cost:.2f} USD")
```
